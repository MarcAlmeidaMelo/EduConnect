```python
# operadores : - , + , * , / , **

print (f" a operação {a} - {b} é igual a: {a - b}")

print (f" a operação {a} * {b} é igual a: {a * b}")

print (f" a operação {a} / {b} é igual a: {a / b}")

print (f" a operação {a} ^ {b} é igual a: {a ** b}")
```
